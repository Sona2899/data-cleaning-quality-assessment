```python
# Standardize text columns
text_cols = ['region', 'payment_mode', 'order_status', 'product_category']

for col in text_cols:
    df[col] = df[col].str.strip().str.title()
```

```python
df['order_date'] = pd.to_datetime(df['order_date'], errors='coerce')
```

```python
df['region'] = df['region'].fillna('Unknown')
```

```python
df['region'].value_counts(dropna=False)
```

```python
df['region'] = df['region'].fillna('Unknown')
```

```python
df['payment_mode'] = df['payment_mode'].fillna('Unknown')
df['order_status'] = df['order_status'].fillna('Unknown')

df[['payment_mode', 'order_status']].isna().sum()
```

```python
df['order_date'] = pd.to_datetime(
    df['order_date'],
    dayfirst=True,
    errors='coerce'
)

df['order_date'].isna().sum()
```

```python
invalid_amounts = df[
    df['total_amount'] != (df['quantity'] * df['unit_price'])
]

invalid_amounts.shape
```

```python
# Round amounts to 2 decimals for reporting
df['unit_price'] = df['unit_price'].round(2)
df['total_amount'] = df['total_amount'].round(2)
```

```python
df.to_csv('operational_transactions_cleaned_python.csv', index=False)
```
